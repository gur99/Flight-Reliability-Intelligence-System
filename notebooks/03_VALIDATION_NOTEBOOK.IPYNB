### Creating Sample of the Data

```python
import pandas as pd
from pathlib import Path

processed_dir = Path("../data/processed")
sample_size_per_month = 834

samples = []

for month in range(1, 13):
    file_path = processed_dir / f"2025_{month:02d}_clean.csv"

    df = pd.read_csv(file_path)

    month_sample = df.sample(
        n=min(sample_size_per_month, len(df)),
        random_state=42
    )

    samples.append(month_sample)

sample_df = pd.concat(samples, ignore_index=True)

sample_df.to_csv(
    "../data/samples/flights_2025_sample.csv",
    index=False
)

print(sample_df.shape)
```
